# Configuración de entorno virtual para YOLOv8 (Windows)

Este notebook documenta, paso a paso, cómo aislar `ultralytics` (YOLOv8) en un entorno virtual
separado, para no romper `numpy`/`numba`/`scipy`/`gensim` en tu entorno principal.

**Importante:** los comandos de esta guía se ejecutan en una **terminal de Windows** (cmd o PowerShell),
**no** dentro de una celda de Jupyter. Esto es porque activar un entorno virtual (`activate`) cambia
el proceso de la terminal donde se ejecuta, y una celda de Jupyter no puede modificar de forma
persistente el proceso del kernel que ya está corriendo. Por eso cada paso se muestra como un bloque
de comandos para copiar y pegar en tu terminal, en orden.


## Paso 1 — Crear el entorno virtual

Abre una terminal (cmd o PowerShell) en la carpeta donde tengas tu notebook de YOLOv8 y ejecuta:

```bat
python -m venv venv_yolo
```

Esto crea una carpeta `venv_yolo` con una instalación de Python aislada, sin tocar tu instalación principal.

## Paso 2 — Activar el entorno

- En **cmd**:
```bat
venv_yolo\Scripts\activate.bat
```

- En **PowerShell**:
```powershell
venv_yolo\Scripts\Activate.ps1
```

Sabrás que quedó activo porque verás `(venv_yolo)` al inicio de la línea de comandos.

## Paso 3 — Instalar ultralytics dentro del entorno

Con el entorno ya activado (debes ver `(venv_yolo)` en el prompt), ejecuta:

```bat
pip install ultralytics
```

Esto instala su propia versión de `numpy` (compatible con `torch`/`ultralytics`) **sin afectar**
el `numpy`/`numba`/`scipy`/`gensim` de tu entorno principal, porque queda aislado dentro de `venv_yolo`.

## Paso 4 — Registrar el entorno como kernel de Jupyter

Sigue con el entorno activado y ejecuta:

```bat
pip install ipykernel
python -m ipykernel install --user --name venv_yolo --display-name "Python (venv_yolo)"
```

Esto agrega un nuevo kernel llamado **Python (venv_yolo)** a la lista de kernels disponibles en Jupyter.

## Paso 5 — Seleccionar el kernel en el notebook de YOLOv8

1. Abre `yolov8_segmentacion.ipynb` en Jupyter.
2. Ve al menú **Kernel > Change Kernel** (o **Kernel > Cambiar kernel**).
3. Selecciona **Python (venv_yolo)**.
4. Ya puedes eliminar o dejar comentada la celda `!pip install ultralytics` de ese notebook,
   porque `ultralytics` ya está instalado dentro de ese entorno.

## Paso 6 — Volver a tu entorno normal cuando termines

En la terminal donde activaste `venv_yolo`, simplemente ejecuta:

```bat
deactivate
```

Tus otros notebooks (Teradata, triángulos, PMPM, etc.) seguirán usando tu entorno principal
sin ningún cambio, ya que `venv_yolo` es completamente independiente.
